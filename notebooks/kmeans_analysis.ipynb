# Shopee Product K-Means Analysis

Notebook này trình bày quy trình phân tích dữ liệu sản phẩm Shopee bằng K-Means.

## 1. Import thư viện

```python
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.cluster import KMeans
from sklearn.preprocessing import LabelEncoder, MinMaxScaler
```

## 2. Đọc dữ liệu

```python
df = pd.read_csv('../data/data.csv', encoding='utf-8-sig')
df.head()
```

## 3. Kiểm tra và xử lý dữ liệu

```python
df = df.drop_duplicates()

for column in ['price', 'buyer_count', 'rating']:
    df[column] = pd.to_numeric(df[column], errors='coerce')

df.isnull().sum()
```

## 4. K-Means theo giá bán

```python
price_scaler = MinMaxScaler()
price_scaled = price_scaler.fit_transform(df[['price']])

price_kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
df['price_cluster'] = price_kmeans.fit_predict(price_scaled)

df[['product_name', 'price', 'price_cluster']].head()
```

```python
top_6 = df.sort_values('price', ascending=False).head(6)

plt.figure(figsize=(10, 5))
plt.bar(top_6['product_name'], top_6['price'])
plt.title('Top 6 sản phẩm có giá bán cao nhất')
plt.xlabel('Sản phẩm')
plt.ylabel('Giá bán')
plt.xticks(rotation=20, ha='right')
plt.tight_layout()
plt.show()
```

## 5. K-Means theo lượt mua và đánh giá

```python
segment_features = ['buyer_count', 'rating']

segment_scaler = MinMaxScaler()
segment_scaled = segment_scaler.fit_transform(df[segment_features])

segment_kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
df['segment_cluster'] = segment_kmeans.fit_predict(segment_scaled)

df[['product_name', 'buyer_count', 'rating', 'segment_cluster']].head()
```

```python
plt.figure(figsize=(9, 5))
plt.scatter(df['buyer_count'], df['rating'], c=df['segment_cluster'])
plt.title('K-Means theo lượt mua và đánh giá')
plt.xlabel('Lượt mua')
plt.ylabel('Đánh giá')
plt.grid(alpha=0.3)
plt.show()
```

## 6. K-Means full feature

```python
encoder = LabelEncoder()
df['category_encoded'] = encoder.fit_transform(df['category'])

full_features = ['price', 'buyer_count', 'rating', 'category_encoded']

full_scaler = MinMaxScaler()
full_scaled = full_scaler.fit_transform(df[full_features])

full_kmeans = KMeans(n_clusters=5, random_state=42, n_init=10)
df['full_cluster'] = full_kmeans.fit_predict(full_scaled)

df[['product_name', 'category', 'price', 'buyer_count', 'rating', 'full_cluster']].head()
```

## 7. Nhận xét

- K-Means theo giá giúp chia sản phẩm thành nhóm giá thấp, trung bình và cao.
- K-Means theo lượt mua và đánh giá giúp quan sát nhóm sản phẩm hot, ít người mua hoặc được đánh giá cao.
- K-Means full feature kết hợp nhiều đặc trưng để phân cụm tổng quát hơn.
